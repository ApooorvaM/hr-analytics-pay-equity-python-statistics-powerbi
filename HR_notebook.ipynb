## Hypothesis Testing
#### Null hypothesis: There is no change in salary of male and female employees
#### Alternate hypothesis: There is a change in salary of male and female employees

##### We are checking if change is statiscally significant

```python
male_salary = data[data['base_salary']>0].loc[data['gender'] == 'Male','base_salary']
female_salary = data[data['base_salary']>0].loc[data['gender']=='Female','base_salary']
```

```python
from scipy.stats import ttest_ind
```

```python
t_stat, p_value = ttest_ind(male_salary,female_salary,equal_var = False)
print('t_stat:', t_stat)
print('p_value:',p_value)
```

#### Since male salary is first variable.. so t stat shows that male salary is 3.91 % more than female salary

#### Since p value comes out to be less that 0.05, result is statistically significant. But we need to account for other variables as company, job title, years at company, seniority, location need to be checked. Hence, we will make use of regression analysis.

## Regression Analysis

```python
reg_data = data[['base_salary','gender','YOE','years_at_company','company','title','nation','race','education']]

reg_data.head()
```

```python
import statsmodels.formula.api as smf
```

#### Regression will help to check if gaps in pay between demographic groups—such as gender, race, and education—still exist after accounting for professional factors like work experience, company tenure, job title, and location. While simple averages show raw differences, they do not consider whether people are in different roles or have different levels of experience. By controlling for these key professional factors, regression isolates the remaining wage variation. This helps HR identify unexplained pay gaps, prioritize fair pay reviews, and make data-driven recommendations to ensure workplace equity.

## Checking association of base salary with gender 

```python
model_gender = smf.ols(formula = """ np.log(base_salary) ~ C(gender)""", data = reg_data[reg_data['base_salary']>0]).fit()
print(model_gender.summary())
```

```python
# Builidng model 1 with base salary vs gender with other factors. We are using log as salary data is right skewed

model_1 = smf.ols(formula = """ np.log(base_salary) ~ C(gender) + YOE + years_at_company + C(company) + C(title) 
                    """ , data = reg_data[reg_data['base_salary']>0]).fit()

print(model_1.summary())
```

#### It does not show any significant difference on the basis of gender

## Checking association of base salary with race

```python
model_race = smf.ols(formula = """ np.log(base_salary) ~ C(race)""", data = reg_data[reg_data['base_salary']>0]).fit()
print(model_race.summary())
```

```python
# builidng model 2 accounting for other factors base salary and race accounting all other factors 

model_2 = smf.ols(formula = """ np.log(base_salary) ~ C(race) + YOE + years_at_company + C(company) + C(title)
                    """, data = reg_data[reg_data['base_salary']>0]).fit()

print(model_2.summary())

```

#### There is a considerable pay difference between race and base salary even after considering other controlling variables

## Lets check association of base salary with education

```python
model_edu = smf.ols(formula = """ np.log(base_salary) ~ C(education)""", data = reg_data[reg_data['base_salary']>0]).fit()
print(model_edu.summary())
```

```python
model_3 = smf.ols(formula = """ np.log(base_salary) ~ C(education) + YOE + years_at_company + C(title) + C(company)""", data = reg_data[reg_data['base_salary']>0]
                  ).fit()

print(model_3.summary())
```

#### Master's and PhD holders have significantly higher estimated salaries than bachelor's degree holders.
#### Some college has a smaller but statistically significant advantage.
#### High school is not statistically significant at the 5% level (p = 0.075).
#### The reduction in education coefficients after adjustment suggests that job allocation, experience and company differences contribute to observed salary variation.

#### Nation and company were overlapping information .. so it was giving singular matrix warning.. so I have removed company in this model

```python
# building model 4 with all factors

model_4 = smf.ols(formula = """ np.log(base_salary) ~ C(nation) + YOE + years_at_company + C(title) 
                     """ , data = reg_data[reg_data['base_salary']>0]).fit()

print(model_4.summary())
```

Important parameters in output:
1\. R-squared (R²): Shows how much variation in annual compensation is explained by the variables included in the model.

2\. Coefficient (coef): Shows the direction and size of a variable's estimated effect on compensation (in percentage terms when using log salary).

3\. P-value: Indicates whether the estimated relationship is statistically significant (commonly, below 0.05).

4\. Confidence Interval (0.025–0.975): Shows the range within which the true coefficient is estimated to lie with 95% confidence.

5\. F-statistic / Prob (F-statistic): Tests whether the regression model as a whole is statistically significant.

6\. No. Observations: Shows the number of records actually used to build the regression model.

## HC3 validation

#### Since traditional ols method assumes equal variance and this data has unequal variance. HC3 provides a more reliable way to calculate statistical uncertainty when variance is unequal. SO we are running HC3 to validate the findings

```python
model_1_hc3 = model_1.get_robustcov_results(cov_type='HC3')

print(model_1_hc3.summary())
```
